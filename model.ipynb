# Simple Dataset

```python
import torch
from torch.utils.data import Dataset
torch.manual_seed(1)
```

### Simple dataset

```python
# Define class for the dataset
class toy_set(Dataset):

    # Constructor with defult values
    def __init__(self, length=100, transform = None):
        self.len = length
        self.x = 2 * torch.ones(length,2)
        self.y = torch.ones(length,1)
        self.transform = transform

    # Getter
    def __getitem__(self, index):
        sample = self.x[index], self.y[index]
        if self.transform:
            sample = self.transform(sample)
        return sample

    # Get length
    def __len__(self):
        return self.len
```

```python
# create dataset object
dataset = toy_set()
print('Our dataset object:', dataset)
print('Value on index 0 of our dataset:', dataset[0])
print('Length of our dataset:', len(dataset))
```

```python
# Print out the first 3 elements in dataset
for i in range(3):
    x, y = dataset[i]
    print(f'Index {i}: x = {x}, y = {y}')
```

```python
for x,y in dataset:
    print(f'x = {x}, y = {y}')
```

```python
# Create a dataset of length 50
my_dataset = toy_set(length=50)
print('Length of my dataset:', len(my_dataset))
```

### Transforms

```python
# We will try to add 1 to x and multiply y by 2
# Create transform class
class my_transform(object):
    # Constructor
    def __init__(self, add = 1, multip = 2):
        self.add = add
        self.multip = multip

    # Executor
    def __call__(self, sample):
        x, y = sample
        x += self.add
        y *= self.multip
        return x, y
```

```python
# Create a Transform object
transform = my_transform()
dataset = toy_set()
```

```python
# Print out first 10 elements in dataset
for i in range(10):
    x, y = dataset[i]
    print(f'Index {i}: x = {x}, y = {y}')
    print('Applying transform to the dataset...')
    x_, y_ = transform((x,y))
    print(f'Index {i}: x_ = {x_}, y_ = {y_}')
```

```python
# Create a new dataset object with the transform applied
transformed_dataset = toy_set(transform=transform)
```

```python
# Print out the first 10 elements in dataset
for i in range(10):    
    x, y = dataset[i]
    print(f'Index {i}: x = {x}, y = {y}')
    x_, y_ = transformed_dataset[i]
    print(f'Index {i}: x_ = {x_}, y_ = {y_}')
```

### Compose

```python
from torchvision import transforms
# Craete a new transform class that multiple each of the elements by 100
class my_transform_2(object):
    # Constructor
    def __init__(self, multip = 100):
        self.multip = multip

    # Executor
    def __call__(self, sample):
        x, y = sample
        x *= self.multip
        y *= self.multip
        return x, y
```

```python
# Combine the two transforms using Compose
composed = transforms.Compose([my_transform(), my_transform_2()])
print(composed)
```

```python
composed(dataset[0])
```

```python
x,y = dataset[0]
x_,y_ = composed(dataset[0])
print(f'Original: x = {x}, y = {y}')
print(f'Transformed: x_ = {x_}, y_ = {y_}')
```

```python
composed_dataset = toy_set(transform=composed)
for i in range(3):
    x, y = dataset[i]
    print(f'Index {i}: x = {x}, y = {y}')
    x_, y_ = composed_dataset[i]
    print(f'Index {i}: x_ = {x_}, y_ = {y_}')
```
